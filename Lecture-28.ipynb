### Sınıfların Static Metotları ve Sınıf Metotları

Bazı fonksiyonlar bir sınıfın konusuyla ilgili işlemler yaparlar ancak herhangi bir sınıf örnek özniteliğini kullanmazlar. yani bunların yalnızca mantıksal bakımdan sınıf ile ilişkileri vardır. Örneğin date isimli bir sınıf tarih işlemleri yapıyor olsun.

class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year

bir yılın artık yıl olup olmadığını tespit eden bir fonksiyon yazmak isteyelim. Bu fonksiyonun aslında tarih işlemleriyle yalnızca mantıksal bir ilgisi vardır. Böyle bir fonksiyon global düzeyde yazılabilir. Ancak bu mantıksal ilgiden dolayı bu foknisoyunn Date sınıfı içerisinde bir metot biçiminde oluşturulması tercih edilir. İşte böylece bir sınıfa dahil olan fakat o sınıfın hiçbir örnek özniteliğini kullanmayan bir metotla karşılaşmış olur. Böyle metotlar "static metot" olarak isimlendirilmektedir. 

static metotların bir self parametresi olmaz. bunlar yalnızca mantıksal bakımdan sınıfla ilişikilidirler. static metotlar @staticmethod isimli bir dekoratörle oluşturulmaktadır.

Artık Yıl
--------------

4 ile eşit bölünebilen herhangi bir yıl artık yıldır: örneğin, 1988, 1992 ve 1996 artık yıllardır. Ancak, yine de hesaba katılması gereken küçük bir hata vardır. Bu hatayı ortadan kaldırmak için, Gregoryen takvimi, 100 ile eşit olarak bölünebilen bir yılın (örneğin, 1900) artık yıl olduğunu, ancak 400 ile de eşit olarak bölünebilir olmasını şart koşar.

Bu nedenle, aşağıdaki yıllar artık yıl değildir:

1700, 1800, 1900, 2100, 2200, 2300, 2500, 2600

Bunun nedeni, 100 ile eşit olarak bölünebilmeleri, ancak 400 ile bölünememeleridir.

Aşağıdaki yıllar artık yıllardır: 1600, 2000, 2400

Bunun nedeni, 100 ve 400 ile eşit olarak bölünebilmeleridir.

```python
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    
    def disp(self):
        print('{}/{}/{}'.format(self.day, self.month, self.year))
    
    @staticmethod
    def is_leap_year(year):
        return year % 400 == 0 or year % 4 == 0 and year % 100 != 0

result = Date.is_leap_year(1800)
print('Artık' if result else 'Artık değil')
```

```python
date = Date(10, 12, 2007)

date.disp() # örnek metodun çağrılması için asıl biçim
```

```python
Date.disp(date) # örnek metodun çağrılması için alternatif biçim
```

```python
result = Date.is_leap_year(2000) # static metodun çağrılması için asıl biçim
print('Artık' if result else 'Artık değil')
```

```python
result = date.is_leap_year(2000) # static metodun çağrılması için alternatif biçim
print('Artık' if result else 'Artık değil')
```

Statik bir metodu bir nesneyle çağırdığımızda sanki o metot o nesnenin özniteliklerini kullanacak gibi yanlış bir algı oluşmaktadır. Aslında yorumlayıcı metodun o nesnenin türüne ilişkin sınıf ismiyle çağrıldığını varsaymaktadır.

s bir sınıf türünden nesne ve foo da static bir metot olmak üzere:

s.foo()

ifadesi ile,

type(s).foo()

ifadesi eşdeğerdir.

```python
class Sample:
    def foo(self):
        print('foo')
    
    @staticmethod
    def bar(self): # tabi bu örnekte bar metodundaki parametrenin self ibiçiminfe isimlendirilmesi kötü bir örnektir.
        print('bar')
    
s = Sample()
s.foo()     # Sample.foo(s)
```

```python
s.bar(10) # geçerli 10 burada self'e atanacak
```

```python
s.bar() # exception, çünkü self sıradan bir parametre
```

Statik metotların yanı sıra python'da aynı zamanda bir de sınıf metotları (class methods) kavramı vardır. Sınıf metotları mantıksal bakımdan static metotlara çok benzerdir. Sınıf metotları @classmethod dekoratörüyle oluşturulmaktadır. Sınıf metotları static metotlardan fakrlı olarak zorunlu bir ilk parametreye sahiptir. Bu ilk parametreye geleneksel olarak cls ismi verilmektedir.

```python
class Sample:
    def __init__(self): # örnek metodu
        self.a = 10
    
    @staticmethod
    def bar():          # statik metot
        print('bar')
    
    @classmethod
    def tar(cls):       # sınıf metodu
        print('tar')
```

Sınıf metotlarının cls parametresi o sınıf metodu hangi sınıfla ilişkinse o sınıfın type nesnesini almaktadır. cls sınıfına neden gereksinim duyulabilir? Bazı uygulamalarda bu parametreye gereksinim duyabiliriz. 

```python
class Sample:
    def __init__(self): # örnek metodu
        self.a = 10
    
    def foo(self):      # örnek metodu
        print('foo')
    
    @staticmethod
    def bar():          # statik metot
        print('bar')
    
    @classmethod
    def tar(cls):       # sınıf metodu
        print('tar')
```

```python
s = Sample()
s.foo() # örnek metodu çağrılır
```

```python
Sample.bar()
```

```python
Sample.tar()
```

```python
s.bar()
```

```python
s.tar() # sample cls parametresi olarak geçirilir.
```

## Dekoratörler

"Fonksiyon ve metot dekoratörleri" ve "sınıf dekoratörleri" olmak üzere ikiye ayrılmaktadırlar. Dekoratörleri belirtmek için "@" karakteri kullanılır. "@" karakterine bitşik olan isme dekoratör denilmektedir.

@foo
@classmethod

birer dekoratördür.

### Fonksiyon ve Metot dekoratörleri



```python
def foo():
    print('foo')

foo()
```

```python
def foo():
    print('other foo')

foo()
```

```python
@foo
def bar():
    print('bar')
```

Bildirimini gören python yorumlayıcısı bu bildirimi aşağıdakiyle eşdeğer kabul etmektedir.
===

def bar():
    print('bar')

bar = foo(bar)

1) Burada bar metodu foo metoduna argüman yapılarak foo metodu çağrılmaktadır. Bu çağırma yalnızca bildirimin bulunduğu yerde bir kez yapılır.

2) Bu çağrının geri dönüş değeri yeniden bar'a atanmatakdır. Böylece artık biz bar fonksiyonunu çağırdığımızda aslında foo fonksiyonun geri verdiği fonksiyonu çağırmış oluruz. Genellikle dekoratör fonksiyonları bir şeyler ypaarak parametresiyle aldığı fonksiyonun aynısına geri dönmektedir. Böylece örneğimizde aslında biz bar metodunu çağırdığımızda yine bar metodu çalışacaktır. Fakat ondan önce foo'daki işlemler de yalnızca bir kez yapılmış olacaktır.

```python
def foo(f):
    print('foo')
    return f

@foo
def bar():
    print('bar')

# eşdeğeri bar = foo(bar)

bar()
```

```python
class Sample:
    @staticmethod
    def foo():
        print('static foo method')

# bu işlemin eşdeğeri şöyledir:

class Sample:
    def foo():
        print('static foo method')
    
    foo = staticmethod(foo)

Sample.foo()
```

Pekiyi dekoratörler neden kullanırlar? İşte bazen bir metot çağrıldığında araya başka bir metodu sokmak gerekebilmektedir. Araya sokulan bu metot (yani dekoratör metodu) bazı ekstra işlemler yapıp bize aynı ya da farklı bir metodu verilir.

Eğer asıl fonksiyonu her çağırdığımızda yeniden başka bir işlemin yapılmasını istiyorsak dekoratör fonksiyonunda biz başka bir fonksiyona geri dönmeliyiz. O fonksiyon da asıl fonksiyonu çağırmalıdır. Bu durum tipik olarak iç fonksiyonlarla sağlanabilir

```python
def foo(f):
    def g():
        print('araya girilen kod')
        f()
    
    return g

@foo
def bar():
    print('bar')

bar()
bar()
```

```python
def foo(f):
    print('araya girilen kod')
    return f

@foo
def bar():
    print('bar')

bar()
bar()
```

Biz dekore edilmiş fonksiyonu çağırdığımızda aslında dekoratör fonksiyonunu çağırmış oluruz. O da dolaylı olarak birşeyler yaptıktan sonra dekore edilmiş fonksiyonu çağırmaktadır. 

Örneğin bir fonksiyon çağrıldığında aynı zamanda bir log tutulmasını isteyelim. Bunu basit bir biçimde dekoratörle yapabiliriz.

```python
import datetime as dt

def mylog(f):
    file = open('logfile.txt', 'w')
    def log():
        print('log')
        file.write('function called at ' + str(dt.datetime.now()) + '\n')
        f()
    
    return log

@mylog
def foo():
    print('foo')

foo()
foo()
foo()
```

@bar('ali', 'veli', 'selami')
def foo():
    pass

eşdeğeri -->

def foo():
    pass

foo = bar('ali', 'veli', 'selami')(foo)

burada dekoratör fonksiyonnu verilen argümanlarla çağrıldığına, onun geri dönüş değeri olarak elde edilen fonksiyona da dekora edilen fonksiyonun verildiğine dikkat edelim. yani bu durumda bir akdae daha çağırma uygulanıyor.

```python
def bar(a, b, c):
    print(f'bar çağrıldı: {a}, {b}, {c}')
    def tar(f):
        print(f'tar çağrıldı: {a}, {b}, {c}')
        def zar():
            print(f'araya girecek kod: {a}, {b}, {c}')
            f()
        return zar
    return tar

@bar('ali', 'veli', 'selami')
def foo():
    print('foo çağrıldı')

foo()
foo()
foo()
```

1) bar fonksiyonu dekoratör parametreleriyle çağrılır. bar fonksiyonu bize geri dönüş dğeri olarak tar fonksiyonunu verir.
2) tar fonksiyonuna foo fonksiyonu argüman oalrak geçirilip geri dönüş değeri foo'ya atanmıştır. yani tar fonksiyonunun geri dönüş değeri olan zar aslında foo çağrıldığında çağrılacak fonksiyon olmaktadır.
3) artık foo çağrıldığında zar fonksiyon çağrılacaktır. bar, tar ve zar fonksiyonları da dekoratör parametrelerile erişip onları kullanılabilecektir.
